# Creating Training Dataset
Creating a target column for each discipline: SL, GS, and SG. Target values are between 0-6:

* 0: Peak World Ranking below 100 in the discipline
* 1: Peak World Ranking between 101-250 in the discipline
* 2: Peak World Ranking between 251-500 in the discipline
* 3: Peak World Ranking between 501-750 in the discipline
* 4: Peak World Ranking between 751-1000 in the discipline
* 5: Peak World Ranking between 1001-1500 in the discipline
* 6: Peak World Ranking over 1500 in the discipline

```python
import pandas as pd

df = pd.read_parquet(
    "../data/processed/training_dataset.parquet"
)

print(df.shape)

# print(df.columns)

display(df)
```

## Visualizing Distribution by Target Category
Plotting how many athletes target label is in each category by discipline

```python
import matplotlib.pyplot as plt

import numpy as np
import matplotlib.pyplot as plt


def plot_target_by_gender(df, target, title):

    counts = (
        df.groupby([target, "Gender"])
        .size()
        .unstack(fill_value=0)
    )

    counts = counts[["M", "F"]]

    print(f"\n{target}:")
    print(counts)

    x = np.arange(len(counts))
    width = 0.35

    plt.figure(figsize=(8, 5))

    plt.bar(
        x - width / 2,
        counts["M"],
        width,
        label="Male",
        color="blue"
    )

    plt.bar(
        x + width / 2,
        counts["F"],
        width,
        label="Female",
        color="pink"
    )

    plt.title(title)
    plt.xlabel("Target Variable")
    plt.ylabel("Count")
    plt.xticks(x, counts.index)
    plt.legend()

    plt.show()


# SL
plot_target_by_gender(
    df,
    "SL_target",
    "Target Distribution for SL by Gender"
)

# GS
plot_target_by_gender(
    df,
    "GS_target",
    "Target Distribution for GS by Gender"
)

# SG
plot_target_by_gender(
    df,
    "SG_target",
    "Target Distribution for SG by Gender"
)


## Plot target by YOB
def plot_target_by_yob(df, target, title):

    counts = (
        df.groupby([target, "YOB"])
        .size()
        .unstack(fill_value=0)
    )

    counts = counts[
        [1997, 1998, 1999, 2000, 2001, 2002,
         2003, 2004, 2005, 2006, 2007]
    ]

    print(f"\n{target}:")
    print(counts)

    x = np.arange(len(counts.index))
    width = 0.07

    plt.figure(figsize=(12, 6))

    # Create one bar for each YOB
    for i, yob in enumerate(counts.columns):
        plt.bar(
            x + (i - 5) * width,
            counts[yob],
            width,
            label=str(yob)
        )

    plt.title(title)
    plt.xlabel("Target Variable")
    plt.ylabel("Count")

    plt.xticks(
        x,
        counts.index
    )

    plt.legend(
        title="YOB",
        bbox_to_anchor=(1.05, 1),
        loc="upper left"
    )

    plt.tight_layout()
    plt.show()

# SL
plot_target_by_yob(
    df,
    "SL_target",
    "Target Distribution for SL by YOB"
)

# GS
plot_target_by_yob(
    df,
    "GS_target",
    "Target Distribution for GS by YOB"
)

# SG
plot_target_by_yob(
    df,
    "SG_target",
    "Target Distribution for SG by YOB"
)
```

As we can see, the data is very skewed. There are very few athletes who reach the lower world rankings. This will have to be taken into consideration later.
